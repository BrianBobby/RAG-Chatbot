```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document

import copy
import re
```

```python
# Load the document
loader = PyPDFLoader("Citi_Business_client Manual.pdf")
pages_pdf = loader.load()
```

```python
pages_pdf
```

```python
pages = copy.deepcopy(pages_pdf)
```

```python
# Clean PDF artifacts (headers, page numbers, extra whitespace)
cleaned_pages = []
for doc in pages:
    text = doc.page_content

    # Remove repeated headers / footers
    text = re.sub(r"CitiBusiness®\s+Client Manual.*?\n", "", text)
    text = re.sub(r"\n\d+\s*$", "", text)  # page numbers

    # Normalize whitespace but keep paragraphs
    text = "\n".join(
        line.strip()
        for line in text.splitlines()
        if line.strip()
    )

    cleaned_pages.append(text)
```

```python
# Merge all pages into one document
full_text = "\n\n".join(cleaned_pages)
```

```python
# Split by MAJOR SECTION HEADERS (e.g. "1. General Information")
section_pattern = re.compile(
    r'(?m)^(\d+\.\s+[A-Z][^\n]+)'
)

matches = list(section_pattern.finditer(full_text))

sections = []
for i, match in enumerate(matches):
    start = match.start()
    end = matches[i + 1].start() if i + 1 < len(matches) else len(full_text)

    section_title = match.group(1).strip()
    section_text = full_text[start:end].strip()

    sections.append(
        Document(
            page_content=section_text,
            metadata={"section": section_title}
        )
    )
```

```python
# Recursive splitting INSIDE each section
splitter = RecursiveCharacterTextSplitter(
    chunk_size=800,        # manuals need larger chunks
    chunk_overlap=120,
    separators=["\n\n", "\n", ". ", " ", ""]
)

chunks = splitter.split_documents(sections)
```

```python
def is_noise_chunk(doc):
    text = doc.page_content.strip().lower()

    # Too short to be meaningful
    if len(text) < 200:
        return True

    # Table of contents / appendix listings
    toc_keywords = [
        "appendix",
        "definitions",
        "disclosure document",
        "contact us"
    ]
    if any(k in text for k in toc_keywords) and "purpose:" not in text:
        return True

    # Looks like pure navigation (page numbers, no sentences)
    if text.count('.') <= 2 and re.search(r'\b\d+\s*$', text):
        return True

    return False


normalized_clean_chunks = []
for doc in chunks:
    # skip noise first (use original doc for noise detection)
    if is_noise_chunk(doc):
        continue

    # normalize newlines to single spaces, trim and collapse multiple spaces
    text = doc.page_content
    # replace any sequence of whitespace-newline-whitespace with a single space
    text_no_newlines = re.sub(r'\s*\n+\s*', ' ', text)
    # collapse multiple spaces to single space
    text_no_newlines = re.sub(r' {2,}', ' ', text_no_newlines).strip()

    # create a new Document with the cleaned text but preserve original metadata
    normalized_doc = Document(page_content=text_no_newlines, metadata=doc.metadata)
    normalized_clean_chunks.append(normalized_doc)

clean_chunks = normalized_clean_chunks
```

```python
# -------------------------
# Embedding + Pinecone upsert (copy into a notebook cell)
# -------------------------
import os
import hashlib
from dotenv import load_dotenv
from typing import List
from tenacity import retry, wait_exponential, stop_after_attempt

# LangChain OpenAI embeddings
from langchain_openai import OpenAIEmbeddings

# Pinecone
import pinecone
from time import sleep

# Load .env (make sure OPENAI_API_KEY, PINECONE_API_KEY, PINECONE_ENV are set there)
load_dotenv()
```

```python
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
PINECONE_API_KEY = os.getenv("PINECONE_API_KEY")
PINECONE_ENV = os.getenv("PINECONE_ENV")  # e.g. "us-west1-gcp"
INDEX_NAME = os.getenv("PINECONE_INDEX", "citibusiness-manual")

if not OPENAI_API_KEY:
    raise RuntimeError("OPENAI_API_KEY not found in environment (.env).")
if not PINECONE_API_KEY:
    raise RuntimeError("PINECONE_API_KEY not found in environment (.env).")
```

```python
# Settings: change as needed
# -------------------------
EMB_MODEL = "text-embedding-3-small"   # prototype: small; switch to text-embedding-3-large for prod
BATCH_EMBED = 64                       # batch size for embedding
BATCH_UPSERT = 64                      # batch size for Pinecone upsert

# Initialize embeddings (OpenAI key is read from env by the lib)
os.environ["OPENAI_API_KEY"] = OPENAI_API_KEY
emb = OpenAIEmbeddings(model=EMB_MODEL)
```

```python
# -------------------------
# Retry wrapper for embedding batches (robust to 429s / transient errors)
# -------------------------
@retry(wait=wait_exponential(multiplier=1, min=1, max=10), stop=stop_after_attempt(5), reraise=True)
def embed_batch(batch: List[str]) -> List[List[float]]:
    """
    Embed a single batch of texts, automatically retrying on transient errors.
    """
    return emb.embed_documents(batch)

def embed_texts_in_batches(texts: List[str], batch_size: int = BATCH_EMBED) -> List[List[float]]:
    """
    Embed a list of texts in batches using embed_batch with retries.
    """
    vectors = []
    for i in range(0, len(texts), batch_size):
        batch = texts[i : i + batch_size]
        batch_vectors = embed_batch(batch)
        vectors.extend(batch_vectors)
    return vectors
```

```python
# Prepare data for embedding
# Assumes `clean_chunks` is already present (list of langchain Document or similar)
# We'll create stable ids using a sha256 of the chunk text
# -------------------------
# If your clean_chunks are plain dicts, adapt accordingly.
texts = []
ids = []
metadatas = []

for i, doc in enumerate(clean_chunks):
    # doc may be a langchain Document object
    text = getattr(doc, "page_content", None) or getattr(doc, "content", None) or doc
    # create stable id from sha of content + index
    sha = hashlib.sha256(text.encode("utf-8")).hexdigest()
    doc_id = f"chunk-{i}-{sha[:8]}"
    texts.append(text)
    ids.append(doc_id)
    # metadata: keep at least section and preview
    md = getattr(doc, "metadata", {}) if hasattr(doc, "metadata") else {}
    md2 = dict(md) if md else {}
    md2.setdefault("source", os.path.basename("Citi_Business_client Manual.pdf"))
    md2["text_preview"] = text[:300]
    md2["sha256"] = sha
    metadatas.append(md2)

print(f"Prepared {len(texts)} chunks for embedding.")
```

```python
texts
```

```python
# -------------------------
# Embed all texts (this may take a few minutes depending on number of chunks)
# -------------------------
vectors = embed_texts_in_batches(texts, batch_size=BATCH_EMBED)
print(f"Computed {len(vectors)} embeddings (dim={len(vectors[0]) if vectors else 'n/a'}).")
```

```python
# Initialize Pinecone and create index if missing
# (This uses the standard pinecone-client pattern)
# -------------------------

from pinecone import Pinecone, ServerlessSpec
pc = Pinecone(
    api_key=PINECONE_API_KEY
)
```

```python
# 3️⃣ Create index if missing
if INDEX_NAME not in pc.list_indexes().names():
    print(f"Creating Pinecone index '{INDEX_NAME}'...")
    pc.create_index(
        name=INDEX_NAME,
        dimension=dim,
        metric="cosine",
        spec=ServerlessSpec(
            cloud="aws",
            region="us-east-1"   # must match your Pinecone console
        )
    )
else:
    print(f"Index '{INDEX_NAME}' already exists.")

# 4️⃣ Connect to index
index = pc.Index(INDEX_NAME)
```

```python
# Upsert into Pinecone in batches
# -------------------------
# Prepare upsert tuples: (id, vector, metadata)
items = [(ids[i], vectors[i], metadatas[i]) for i in range(len(ids))]

for i in range(0, len(items), BATCH_UPSERT):
    batch = items[i : i + BATCH_UPSERT]
    print(f"Upserting batch {i//BATCH_UPSERT + 1} / { (len(items)-1)//BATCH_UPSERT + 1 } ...")
    # Pinecone expects list of dicts or tuples; tuple (id, vector, metadata) works
    index.upsert(vectors=batch)
    # small sleep to avoid any burst throttling on Pinecone side
    sleep(0.1)

print("All chunks upserted to Pinecone ✅")
```

```python
# Simple retrieval function (sanity check)
# -------------------------
def retrieve(query: str, top_k: int = 5):
    q_emb = emb.embed_query(query)
    resp = index.query(vector=q_emb, top_k=top_k, include_metadata=True)
    # resp may be a dict or an object depending on pinecone client version
    matches = resp.get("matches", []) if isinstance(resp, dict) else getattr(resp, "matches", [])
    return matches

# Example test
q = "What are identity verification requirements?"
matches = retrieve(q, top_k=5)
print("Top matches (id - score - section):")
for m in matches:
    meta = m.get("metadata", {}) if isinstance(m, dict) else m.metadata
    score = m.get("score", None) if isinstance(m, dict) else m.score
    print(m["id"] if isinstance(m, dict) else m.id, round(score, 4) if score else score, meta.get("section"))
    # optionally print preview:
    print(" preview:", meta.get("text_preview")[:200])
    print()
```
